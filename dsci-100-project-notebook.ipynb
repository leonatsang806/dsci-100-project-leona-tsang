#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

```r
#BRAINSTORM
#omit na 
#select numerical vars
#summarise (across, functions)
####################################################################################################################



summary_march_madness <- march_madness |>
    na.omit()|>  #Removing all rows containing NA as they will interfere with the mean, min, and max functions when I summarise my data (any NA observation will result in the overall function returning NA)
    select(-school,-conference, -division, -bid,-tourney_finish, -first_game_at_home)|> #Removing all non-numerical columns using the select function
    summarise(across(year:total_wins_pct, c(max=max, min=min, mean=mean))) 

#I am using the summarise function to create a new data frame with a max, min, and mean column for each numerical variable in the march_madness data frame.
#Summarise is the most suitable function as I am collecting aggregate statistics of my data set. 
#The across function inside summarise allows me to perform my list of aggregrate statistic functions across each column in the range specified

summary_march_madness            
```

```r
#I am using a histogram to display the distribution of regional win percentages across my tidied data set. 
#Histograms are the most suitable visualisation for displaying the distribution of a single numerical variable 

options(repr.plot.width=10, repr.plot.height=10)
reg_win_pct_distribution <- tidy_march_madness |> 
    ggplot(aes(x=reg_wins_pct))+ 
    geom_histogram(binwidth=4) +
    labs(x='Regional Win Percentage (%)', 
        y='Number of Records',
        title='Distribution of Regional Win Percentages (%) of NCAA Division I\n Women’s Basketball Tournament Records from 1982 to 2018')+
    theme(text=element_text(size=18))
reg_win_pct_distribution
```

```r
#I am creating a barchart to compare the relative proportions of each type of bid (auto vs at-large) in my data set 
#Bar charts are the most suitable visualisation for comparing amounts/proportions between categorical variables

#First, I need to create a data frame with two columns: one representing the type of bid, and one representing the relative proportion of that bid in the data set.
#This way I can easily map the relative proportions of each bid to my ggplot function later. 
bid_props <- tidy_march_madness |>
    group_by(bid) |>
    summarise(prop=n()) |>
    mutate(prop=prop/nrow(tidy_march_madness))
bid_props

#This bar chart will compare the relative proportions of auto vs at-large bids in the NCAA record data set 
options(repr.plot.width=10, repr.plot.height=10)
bid_distribution <- bid_props |>
    ggplot(aes(x=bid, y=prop, fill=bid)) +
    geom_bar(stat='identity') +
    labs(x='Type of bid', 
         y='Relative Proportion of NCAA records',
        title='Relative Proportion of auto vs at-large bids in NCAA Division I\n Women’s Basketball Tournament Records') +
    theme(text=element_text(size=20))+
    scale_y_continuous(breaks = seq(0, 1, 0.05))

bid_distribution
```

**Visualization 1: A histogram comparing the distribution of conference win percentages between at-large and auto bids**

This histogram helps me visualise the distribution of conference win percentages for at-large and auto bids. This may prove useful when interpreting the how the conference win percentage distribution of each class may affect the distance calculation during k-nn classification. For example, the histogam shows at-large bids having a more even distribution of conference win percentages than auto bids, so new observations with lower conference win percentages may be more likely to be classified at 'at-large'. 

```r

options(repr.plot.width=12, repr.plot.height=12)
bid_reg_win_pct_distribution <- tidy_march_madness |> 
    ggplot(aes(x=conf_wins_pct, fill=bid))+ 
    geom_histogram(binwidth=8) +
    labs(x='Conference Win Percentage (%)', 
        y='Number of NCAA Records',
        title='Distribution of Conference Win Percentages of NCAA Division I\n Women’s Basketball Tournament Records from 1982 to 2018')+
    theme(text=element_text(size=18))+
    facet_grid(rows=vars(bid))
bid_reg_win_pct_distribution
```

**Visualization 2: A histogram comparing the distribution of regional win percentages between at-large and auto bids****

Similar to visualisation 1, using a histogram to visualise the distribution of regional win percentages within a specific bid class can help us quickly gauge what type of bid a new observation is likely to be assigned. This visualisation shows that auto bids have a more even distribution of regional win percentages, while at-large records are mostly concentrated around ~60-80%. Therefore, we would expect our classification model to likely predict observations with more extreme win percentages as an 'auto' class.

```r
options(repr.plot.width=12, repr.plot.height=12)
bid_reg_win_pct_distribution <- tidy_march_madness |> 
    ggplot(aes(x=reg_wins_pct, fill=bid))+ 
    geom_histogram(binwidth=6) +
    labs(x='Regional Win Percentage (%)', 
        y='Number of NCAA Records',
        title='Distribution of Regional Win Percentages of NCAA Division I\n Women’s Basketball Tournament Records from 1982 to 2018')+
    theme(text=element_text(size=18))+
    facet_grid(rows=vars(bid))
bid_reg_win_pct_distribution
```
